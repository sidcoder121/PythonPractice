91. How do you find the length of the longest substring without repeating characters?

```python
s = "abcabcbbbcccdbcccd"
longest_substring = 0
char_set = set()
left = 0

for right in range(len(s)):
    while s[right] in char_set:
        char_set.remove(s[left])
        left += 1
    char_set.add(s[right])
    longest_substring = max(longest_substring, right - left + 1)

print(longest_substring)
```

95. How do you find the longest palindromic substring in a string?

```python
def longest_palindrome(s):
    longest = ""
    for i in range(len(s)):
        for j in range(i + len(longest), len(s) + 1):
            substring = s[i:j]
            if substring == substring[::-1]:
                longest = substring
    return longest

s = "bghsghhdd"
print(longest_palindrome(s))
```

97. How do you check if two strings are one edit distance apart (insert, delete, or replace one character)?

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

len_s1 = len(s1)
len_s2 = len(s2)

if abs(len_s1 - len_s2) > 1:
    print("The strings are NOT one edit distance apart.")
else:

    if len_s1 == len_s2:
        diff_count = 0
        for i in range(len_s1):
            if s1[i] != s2[i]:
                diff_count += 1
                if diff_count > 1:
                    break
        if diff_count == 1:
            print("The strings ARE one edit distance apart (replace).")
        else:
            print("The strings are NOT one edit distance apart.")


    else:

        if len_s1 > len_s2:
            s1, s2 = s2, s1
            len_s1, len_s2 = len_s2, len_s1

        i = 0
        j = 0
        diff_found = False

        while i < len_s1 and j < len_s2:
            if s1[i] != s2[j]:
                if diff_found:
                    print("The strings are NOT one edit distance apart.")
                    break
                diff_found = True
                j += 1
            else:
                i += 1
                j += 1

        if not diff_found or (diff_found and j == len_s2 - 1):
            print("The strings ARE one edit distance apart (insert/delete).")
        else:
            print("The strings are NOT one edit distance apart.")
```

99. How do you find the number of substrings that are palindromes in a string?

```python
s = "abbba"
palindrome_count = 0

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):
        substring = s[i:j]
        if substring == substring[::-1]:
            palindrome_count += 1

print(palindrome_count)
```
