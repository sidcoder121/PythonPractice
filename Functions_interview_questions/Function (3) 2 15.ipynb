5.Write a recursive function to find the nth Fibonacci number.

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

result = fibonacci(10)
print(result)
```

7.Write a closure function that returns a function to multiply by a given factor.

```python
def multiplier(factor):
    def multiply_by_factor(number):
        return number * factor
    return multiply_by_factor

multiply_by_5 = multiplier(5)

print(multiply_by_5(10))
```

8.Write a decorator that logs the arguments and return value of a function.

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Arguments: {args}, {kwargs}")
        result = func(*args, **kwargs)
        print(f"Return value: {result}")
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

add(5, 3)

```

9.Write a higher-order function that applies a function to every element in a list.

```python
def apply_function(data, func):
    return [func(x) for x in data]
result = apply_function([1, 2, 3, 4], lambda x: x**2)
print(result)
```

12.Write a memoized version of the Fibonacci function to improve performance.

```python
def memoize_fib(f):
    memo = {}
    def wrapper(n):
        if n not in memo:
            memo[n] = f(n)
        return memo[n]
    return wrapper

@memoize_fib
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))
```

13.Write a function that checks whether a number is prime.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(30))
```

14.Write a function that uses nested functions to calculate compound interest.

```python
def compound_interest(principal, rate, time):
    def calculate():
        return principal * ((1 + rate) ** time)

    return calculate()

result = compound_interest(1000, 0.05, 2)
print(result)
```

15.Write two functions and create a third function that composes them.

```python
def multiply_by_2(x):
    return x * 2

def add_3(x):
    return x + 3

def compose(f, g):
    return lambda x: f(g(x))

new_function = compose(multiply_by_2, add_3)
print(new_function(5))
```
